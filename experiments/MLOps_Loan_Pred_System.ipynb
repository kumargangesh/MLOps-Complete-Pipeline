```python
import pandas as pd
```

```python
loan = pd.read_csv('/content/train_and_test2.csv')
loan.info()
```

```python
new_data = loan.drop(columns=['Passengerid', 'zero', 'zero.1', 'zero.2', 'zero.3', 'zero.4', 'zero.5', 'zero.6', 'zero.7', 'zero.8', 'zero.9', 'zero.10', 'zero.11', 'zero.12', 'zero.13', 'zero.14', 'zero.15', 'zero.16', 'zero.17', 'zero.18'])
new_data.info()
```

```python
from sklearn.model_selection import train_test_split

trainx, testx, trainy, testy = train_test_split(new_data.drop(columns='2urvived'), new_data['2urvived'], test_size=0.25, random_state=42)
```

```python
from xgboost import XGBClassifier

xgbc = XGBClassifier()
xgbc.fit(trainx, trainy)

xgbc_pred = xgbc.predict(testx)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(testy, xgbc_pred)
```
